```python
import os
#os.mkdir("image_patches/yolov8_train_annotations")
class_map = {
    "Rod": 0,
    "Dividing": 1,
    "Microcolony": 2
}
```

```python
import xml.etree.ElementTree as ET
def xml_to_txt(root):
    file = []
    image_width = int(root.find("size/width").text)
    image_height = int(root.find("size/height").text)

    for obj in root.findall("object"):
        name = obj.find("name").text
        name_id = class_map[name]
        counts[name_id] +=1
        xmin = int(obj.find("bndbox/xmin").text)
        ymin = int(obj.find("bndbox/ymin").text)
        xmax = int(obj.find("bndbox/xmax").text)
        ymax = int(obj.find("bndbox/ymax").text)

        x_center = (xmin+xmax)/2 / image_width
        y_center = (ymin+ymax)/2 / image_height
        width = (xmax - xmin) / image_width
        height = (ymax - ymin) / image_height
        result = [name_id, x_center, y_center, width, height]
        for i in range(5): result[i] = str(result[i])
        result = (" ").join(result)
        file.append(result)
    return ("\n").join(file)
```

```python
folder_path = "augment_dataset2/labels/test_annotations"
xml_files = os.listdir(folder_path)
counts = [0, 0, 0]
for x in xml_files:
    tree = ET.parse("augment_dataset2/labels/test_annotations/" + x)
    root = tree.getroot()
    txt = xml_to_txt(root)
    path = "augment_dataset2/labels/test/" + x[:-3] + "txt"
    with open(path, "w") as f:
        f.write(txt)
print(counts)
```

```python
import os
import shutil

def move_png_files(source_folder, destination_folder):
    """
    Moves all PNG files from a source folder to a destination folder.

    Args:
        source_folder (str): The path to the folder containing the PNG files.
        destination_folder (str): The path to the folder where PNG files will be moved.
    """
    if not os.path.exists(source_folder):
        print(f"Error: Source folder '{source_folder}' does not exist.")
        return

    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
        print(f"Created destination folder: '{destination_folder}'")

    for filename in os.listdir(source_folder):
        if filename.lower().endswith('.txt'):  # Check for .png extension (case-insensitive)
            source_path = os.path.join(source_folder, filename)
            destination_path = os.path.join(destination_folder, filename)

            try:
                shutil.move(source_path, destination_path)
                print(f"Moved: '{filename}' to '{destination_folder}'")
            except Exception as e:
                print(f"Error moving '{filename}': {e}")

# Example usage:
source_directory = "path/to/your/source_folder"  # Replace with your source folder path
destination_directory = "path/to/your/destination_folder"  # Replace with your destination folder path

move_png_files(source_directory, destination_directory)
```
