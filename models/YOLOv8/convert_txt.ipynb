```python
import os
#os.mkdir("image_patches/yolov8_train_annotations")
class_map = {
    "Rod": 0,
    "Dividing": 1,
    "Microcolony": 2
}
```

```python
import xml.etree.ElementTree as ET
def xml_to_txt(root):
    file = []
    image_width = int(root.find("size/width").text)
    image_height = int(root.find("size/height").text)

    for obj in root.findall("object"):
        name = obj.find("name").text
        name_id = class_map[name]
        counts[name_id] +=1
        xmin = int(obj.find("bndbox/xmin").text)
        ymin = int(obj.find("bndbox/ymin").text)
        xmax = int(obj.find("bndbox/xmax").text)
        ymax = int(obj.find("bndbox/ymax").text)

        x_center = (xmin+xmax)/2 / image_width
        y_center = (ymin+ymax)/2 / image_height
        width = (xmax - xmin) / image_width
        height = (ymax - ymin) / image_height
        result = [name_id, x_center, y_center, width, height]
        for i in range(5): result[i] = str(result[i])
        result = (" ").join(result)
        file.append(result)
    return ("\n").join(file)
```

```python
folder_path = "augment_dataset2/labels/test_annotations"
xml_files = os.listdir(folder_path)
counts = [0, 0, 0]
for x in xml_files:
    tree = ET.parse("augment_dataset2/labels/test_annotations/" + x)
    root = tree.getroot()
    txt = xml_to_txt(root)
    path = "augment_dataset2/labels/test/" + x[:-3] + "txt"
    with open(path, "w") as f:
        f.write(txt)
print(counts)
```
